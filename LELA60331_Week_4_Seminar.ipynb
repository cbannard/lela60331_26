In linear regression we fit a line to the data that allows us to predict y from x. The equation for this is as follows (note that I am using the machine-learning term bias in place of the term intercept which is more commonly used in statistics).

y = bias + x*weight [+ error]

Let's start by setting a random weight and setting our bias to zero. `random.random()` gives a random number between 0 and 1.

```python
# We will draw the data with a fitted line several times, so we write a function for it
def plot_fit(bias, weight):
    plt.scatter(x, y)
    plt.xlim(0, 5)
    plt.ylim(0, 100)
    line_x = [0, 5]
    line_y = [bias, bias + 5 * weight]      # the line's height at x = 0 and at x = 5
    plt.plot(line_x, line_y, color='red')
    plt.show()

plot_fit(bias, weight)
```

As you can see this line isn't a very good fit to the data. If we used it for predictions we would get very bad estimates.

We will use gradient descent to find the line of best fit. Note: it is called simply gradient descent when we update weights based on the whole data set as we will here. If we update weights based on random subsets of the data (mini-batch training) it is called stochastic gradient descent. These variants will be explained in a forthcoming lecture.

Gradient descent works as follows: \
Initialise weights to e.g. random values \
For a given number of N iterations:
1. Predict each y value given x, bias and the current weight(s), giving us y_hat
2. Calculate the loss (for reporting/monitoring). For linear regression this is the "mean squared error". We calculate a vector of item-specific "errors" by subtracting the vector of real y values from the vector of estimated y_hat values. We then calculate the dot product of this vector with itself (the sum of the squared values) and divide that by the number of data points.
3. Calculate dw - the gradient of the loss function with regard to each weight. For linear regression this is the dot product of the vector of x values for the given feature and the vector of errors, divided by the number of data points in our data.
4. Calculate db - the gradient of the loss function with regard to the bias. For linear regression this is the sum of the vector of errors, divided by the number of data points in our data.
5. Update each weight (in this first example there is only one) by setting it to be the current weight minus dw times the learning rate
6. Update the bias by setting it to be the current bias minus db times the learning rate
7. Repeat until done

In plain Python a **vector** is just a list of numbers. We need the **dot product** of two vectors several times, so here is a function for it. It multiplies the two lists element by element and adds up the results (just like the ⭐ challenge in Exercise 4 of the Week 3 Python workbook).

```python
def dot(a, b):
    # the dot product of two lists of the same length
    total = 0
    for i in range(len(a)):
        total += a[i] * b[i]
    return total

print(dot([1, 2, 3], [4, 5, 6]))     # 1*4 + 2*5 + 3*6 = 32
```

## Logistic Regression

Next we are going to apply what we have learned to logistic regression.

First we will generate some random data for an imagined sentiment classification task with only two features. Our two features are the log of the count of positive words (e.g. good, excellent) and the log of the count of negative words (e.g. bad, rubbish) in each text. The label we are trying to predict is either 1 (positive sentiment text) or 0 (negative sentiment text).

`random.gauss(mu, sigma)` gives a random number from a normal distribution with mean `mu` and standard deviation `sigma`.

```python
## Create simulated data
random.seed(10)
negative_center = (2, 3)       # texts with fewer positive words than negative words
positive_center = (3, 2)       # texts with more positive words than negative words
batch_size = 50

x = [[], []]       # x[0] will hold feature 1 for every text and x[1] will hold feature 2
y = []
for i in range(batch_size):
    if random.random() > 0.5:
        center = negative_center
        label = 0
    else:
        center = positive_center
        label = 1
    x[0].append(random.gauss(center[0], 1))
    x[1].append(random.gauss(center[1], 1))
    y.append(label)
```

We can visualise the data as follows. The stars are the negative sentiment texts (label 0) and the circles are the positive sentiment texts (label 1).

To draw the two groups separately we need to split each feature's values by label, so we write a function that does this and **returns two lists**.

```python
def split_by_label(values):
    # split a list of feature values into those for negative texts (label 0) and those for positive texts (label 1)
    negative = []
    positive = []
    for k in range(len(values)):
        if y[k] == 0:
            negative.append(values[k])
        else:
            positive.append(values[k])
    return negative, positive

x0_neg, x0_pos = split_by_label(x[0])
x1_neg, x1_pos = split_by_label(x[1])
plt.scatter(x0_neg, x1_neg, marker='*', s=100)
plt.scatter(x0_pos, x1_pos, marker='o', s=100)
plt.xlabel("log count of positive words")
plt.ylabel("log count of negative words")
plt.xlim((0, 5))
plt.ylim((0, 5))
plt.show()
```

To see why we might take the log, we can exponentiate the log counts (reversing the log function) to give raw counts. These are worse for visualisation and modelling purposes.

```python
x_exp = [[math.exp(value) for value in x[0]],
         [math.exp(value) for value in x[1]]]
x0_neg, x0_pos = split_by_label(x_exp[0])
x1_neg, x1_pos = split_by_label(x_exp[1])
plt.scatter(x0_neg, x1_neg, marker='*', s=100)
plt.scatter(x0_pos, x1_pos, marker='o', s=100)
plt.xlabel("count of positive words")
plt.ylabel("count of negative words")
plt.xlim((0, 150))
plt.ylim((0, 150))
plt.show()
```

Our goal in logistic regression is to find a line that allows us to estimate a probability that any text has positive sentiment. If that probability is greater than 0.5 then we will say that it is a positive text and if lower then we will say it is a negative text.

In logistic regression we first estimate a value z as a linear function of our predictors, just as in linear regression:

$z_{i} = bias + x1_{i} \cdot weight1 + x2_{i} \cdot weight2$

We then use the sigmoid function to convert this z value to a probability:

$p(y_{i}=1) = \frac{1}{1+e^{-z_{i}}}$

We can start by setting some random weights and an arbitrary bias.

```python
random.seed(10)
num_features = 2
weights = [random.random() for j in range(num_features)]
bias = 0
```

We can add this line to our plot of values. It should cut across the items so that the positive texts are on one side of the line and the negative texts on the other.

```python
# We will draw the data with the decision line twice, so we write a function for it
def plot_decision_line(bias, weights):
    x0_neg, x0_pos = split_by_label(x[0])
    x1_neg, x1_pos = split_by_label(x[1])
    plt.scatter(x0_neg, x1_neg, marker='*', s=100)
    plt.scatter(x0_pos, x1_pos, marker='o', s=100)
    plt.xlim((-5, 5))
    plt.ylim((-5, 5))
    # the line where z = 0, i.e. where the model's probability is exactly 0.5
    c = -bias / weights[1]
    m = -weights[0] / weights[1]
    xd = [0, 5]
    yd = [m * value + c for value in xd]
    plt.plot(xd, yd, 'k', lw=1, ls='--')
    plt.show()

plot_decision_line(bias, weights)
```

Our random line does not do this. So we will use gradient descent to find the line of best fit.

For logistic regression we use a cross entropy loss function. I have included this in the code (See lecture for details).

To calculate the gradient of the loss function with respect to the bias term we first calculate the difference between each predicted y value and the true y value. We then take the average difference by summing the differences and dividing the result by N - the number of data points in our data:

$\delta b = \frac{1}{N} \sum_{i=1}^{N} (q_{i} - y_{i})$

To calculate the gradient of the loss function with respect to each weight, we again first calculate the difference between each predicted y value and the true y value. We then calculate the dot product of this vector and the vector of x values for the relevant feature and divide the result by N - the number of data points in our data:

$\delta w = \frac{1}{N} \sum_{i=1}^{N} x_{i} (q_{i} - y_{i})$

x here is the vector of values for the feature relevant to the individual weight. A different gradient is needed for each weight and this will be calculated using a different x.
