## 2. List comprehensions: doing the same thing to every item

Very often we want to make a **new list** by doing the same thing to every item of an old one. With a loop that takes four lines:

````python
doubled = []
for s in scores:
    doubled.append(s * 2)
print(doubled)
````

A **list comprehension** does exactly the same in one line:

````python
doubled = [s * 2 for s in scores]
print(doubled)
````

Read it from the middle out: *"for each `s` in `scores`, work out `s * 2`, and collect all the answers in a list"*.

```
[ s * 2     for s in scores ]
  ^^^^^     ^^^^^^^^^^^^^^^
  what to   which items to
  calculate go through
```

The variable name (`s` here) is up to you. It only exists inside the brackets.

## 4. Working through two lists together

Many calculations need **two** lists at once, for example the error for each student is *their prediction minus their real score*. Going through `for s in scores` only gives us one list. Instead we go through the **indexes** with `k`, and use `k` to pick the matching item out of each list:

```python
[y_est[k] - scores[k] for k in range(len(scores))]
```

*"For each index `k`, take item `k` of `y_est` minus item `k` of `scores`."* This works because item `k` of each list belongs to the same student.

This "loop over `k`" pattern is the most important idea in this workbook. You will use it in almost every line of the seminar.

**Exercise 4.3**

The **dot product** of two lists multiplies them item by item and adds up the results. For example, the dot product of `[1, 2, 3]` and `[4, 5, 6]` is 1×4 + 2×5 + 3×6 = 32.

Complete the function `dot(a, b)` using a loop with `k`. (We look at functions in more detail in Section 5.)

````python
def dot(a, b):
    total = 0
    for k in range(len(a)):
        total = total + a[k] * b[k]
    return total

print(dot([1, 2, 3], [4, 5, 6]))
print(dot(hours, errors))
````

The **sigmoid** function turns any number $z$ into a number between 0 and 1:

$$\text{sigmoid}(z) = \frac{1}{1 + e^{-z}}$$

## 7. List comprehensions with lists of lists

There are three kinds of job we do with lists of lists in the seminar. Each needs a slightly different comprehension, and the key question is always **what are we going through: the features (`f`), or the items (`k`)?**

| Job | Goes through | Result |
|---|---|---|
| A. Change every value of **one feature** | the values in one row | one new row |
| B. Combine **all the features for each item** | items `k` | one number per item |
| C. Summarise **each feature** | features `f` | one number per feature |

Writing one line per feature gets tedious. Instead we can **loop over the features** with `f`, and replace each row with a new version of itself. The seminar does exactly this when it scales the data:

```python
for f in range(len(x)):

```

**Exercise 7.2**

Complete the loop so that it adds 1 to every value in every row of `counts_plus_one`, then takes the log. (We copy `counts` first so the original stays unchanged.)

````python
counts_plus_one = [row[:] for row in counts]       # a copy of counts

for f in range(len(counts_plus_one)):
    counts_plus_one[f] = [value + 1 for value in counts_plus_one[f]]
print(counts_plus_one)

log_counts = [row[:] for row in counts_plus_one]   # a copy
for f in range(len(log_counts)):
    log_counts[f] = [math.log(value) for value in log_counts[f]]
print([[round(v, 2) for v in row] for row in log_counts])
````

You may have noticed the print line `[[round(v, 2) for v in row] for row in log_counts]`. That is a **nested** list comprehension: a comprehension inside a comprehension. Read it from the outside in: *"for each `row` in `log_counts`, make the list `[round(v, 2) for v in row]`"*. It does the same job as the `for f` loop, but builds a new list of lists instead of changing the old one.

````python
x_scaled = [row[:] for row in x]      # a copy

for f in range(len(x_scaled)):
    mean = statistics.mean(x_scaled[f])
    sd = statistics.pstdev(x_scaled[f])
    x_scaled[f] = [(value - mean) / sd for value in x_scaled[f]]

print([[round(v, 2) for v in row] for row in x_scaled])
print([round(statistics.mean(row), 2) for row in x_scaled])
print([round(statistics.pstdev(row), 2) for row in x_scaled])
````

## 8. Updating values in a loop

Gradient descent changes the same variables again and again. A line like

```python
weight = weight - lr * dw
```

means: *work out the right-hand side using the **current** `weight`, then store the answer as the new `weight`*. This also works for one item of a list: `weights[0] = weights[0] - lr * dw1`.

We often want to remember a value from every round of a loop (for example the loss, so we can plot it). We do that by starting with an empty list and **appending** to it each time.

**Exercise 8.1**

Start with `value = 100`. Write a loop that runs 6 times. Each time, it should append the current value to `history`, then halve `value`.

````python
value = 100
history = []

for i in range(6):
    history.append(value)
    value = value / 2

print(history)
print(value)
````

## 9. Putting it all together: gradient descent by hand

Now we will do the seminar's Activity 1 on a tiny data set where we know the right answer: y is always exactly 2 × x, so the best line has **weight 2 and bias 0**.

**Exercise 9.1: one step**

Fill in each line for **one** step of gradient descent. Each step uses something from earlier in this workbook. The expected output lets you check every number.

````python
x1 = [1, 2, 3]
y1 = [2, 4, 6]
n = len(y1)
weight = 0
bias = 0
lr = 0.1

y_est = [bias + xi * weight for xi in x1]
errors = [y_est[k] - y1[k] for k in range(n)]
loss = dot(errors, errors) / n
dw = dot(x1, errors) / n
db = sum(errors) / n
weight = weight - lr * dw
bias = bias - lr * db

print(y_est)
print(errors)
print(round(loss, 3))
print(round(dw, 3), round(db, 3))
print(round(weight, 3), round(bias, 3))
````

**Exercise 9.2: many steps**

Now put the same lines inside a loop that runs 500 times, appending the loss to `losses` each time. The weight should end up very close to 2 and the bias very close to 0, and the loss should fall towards 0.

````python
weight = 0
bias = 0
losses = []

for i in range(500):
    y_est = [bias + xi * weight for xi in x1]
    errors = [y_est[k] - y1[k] for k in range(n)]
    loss = dot(errors, errors) / n
    losses.append(loss)
    dw = dot(x1, errors) / n
    db = sum(errors) / n
    weight = weight - lr * dw
    bias = bias - lr * db

print(round(weight, 3), round(bias, 3))
print([round(l, 3) for l in losses[:5]])
print(round(losses[-1], 6))
````

**Exercise 9.3 ⭐: two features**

Here the data is a list of lists with two features, and y = 3 × feature 0 − 1 × feature 1 + 2 exactly. Write the gradient descent loop for **two** weights (as in the seminar's Activity 5). After 2000 steps you should be close to weights `[3, -1]` and bias `2`.

````python
x2 = [[1, 2, 3, 4, 5],
      [2, 1, 4, 3, 5]]
y2 = [3, 7, 7, 11, 12]
n = len(y2)
weights = [0, 0]
bias = 0
lr = 0.02

for i in range(2000):
    y_est = [bias + x2[0][k] * weights[0] + x2[1][k] * weights[1] for k in range(n)]
    errors = [y_est[k] - y2[k] for k in range(n)]
    dw1 = dot(x2[0], errors) / n
    dw2 = dot(x2[1], errors) / n
    db = sum(errors) / n
    weights[0] = weights[0] - lr * dw1
    weights[1] = weights[1] - lr * dw2
    bias = bias - lr * db

print([round(w, 2) for w in weights], round(bias, 2))
````
